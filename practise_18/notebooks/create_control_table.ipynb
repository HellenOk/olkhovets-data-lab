```python
from pyspark.sql import types as T

# 1. Створюємо схему control, якщо її ще немає
spark.sql("CREATE SCHEMA IF NOT EXISTS control")

# 2. Видаляємо таблицю перед створенням, щоб повторний запуск повертав її до початкових 3 рядків без дублікатів
spark.sql("DROP TABLE IF EXISTS control.files_to_load")

# 3. Задаємо точну схему колонок згідно з умовою завдання
schema = T.StructType([
    T.StructField("file_name", T.StringType(), True),
    T.StructField("target_schema", T.StringType(), True),
    T.StructField("target_table", T.StringType(), True),
    T.StructField("load_mode", T.StringType(), True),
    T.StructField("is_active", T.BooleanType(), True),
    T.StructField("last_loaded_at", T.TimestampType(), True),
    T.StructField("last_rows", T.LongType(), True),
])

# 4. Початкові дані (last_loaded_at та last_rows мають значення None / NULL)
rows = [
    ("stations.json", "air_quality", "stations", "overwrite", True, None, None),
    ("sensors.json", "air_quality", "sensors", "overwrite", True, None, None),
    ("readings.json", "air_quality", "readings", "overwrite", False, None, None),
]

df = spark.createDataFrame(rows, schema=schema)

# 5. Записуємо у форматі Delta в таблицю control.files_to_load
df.write.format("delta").mode("overwrite").saveAsTable("control.files_to_load")

# 6. Перевіряємо результат
display(spark.table("control.files_to_load"))

```
